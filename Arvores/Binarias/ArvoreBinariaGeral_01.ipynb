```python
"""
UFRJ - Universidade Federal do Rio de Janeiro
CCMN - Centro de Ciência Matemáticas e da Natureza
IM   - Instituto de Matemática

Disciplina: Estrutura de Dados (MAE014 - ED.2026.2)
Prof.     : Milton R Ramirez (milton@matematica.ufrj.br)

Data Criação  : 08 de outubro de 2026
Data Alteração: 08 de outubro  de 2026

Exemplos de implementação de arvore binárias
"""

class NoArvore_binaria:
  def __init__(self, dado):
    self.dado = dado
    self.esq = None
    self.dir = None

class Arvore_binaria:
  def __init__(self):
    self.raiz = None

  def inserir(self, dado, no = None):
    if no is None:
      if self.raiz is None:
        self.raiz = NoArvore_binaria(dado)
      else:
        self.inserir(dado, self.raiz)
      return
    
    else:
      if no.esq is None:
        no.esq = NoArvore_binaria(dado)
      elif no.dir is None:
        no.dir = NoArvore_binaria(dado)
      else:
        self.inserir(dado, no.esq)

  def buscar(self, dado, no = None, nivel = 0):
    if no is None:
      no = self.raiz  #Chama o no de raiz
      nivel = 1       #definimos o nivel inicial da raiz 1
    return self._buscar_recursivo(dado, no , nivel)

  def _buscar_recursivo(self, dado, no, nivel):
    if no is None:
        return None  #Verifica se o no é vazio

    if no.dado == dado: #Compara o no do nosso dado atual, com o dado fornecido
        print("-" * nivel, no.dado, f"(nível {nivel})")
        return no #Retorna o no se o if for true


    encontrado = self._buscar_recursivo(dado, no.esq, nivel+1)
    if encontrado is not None:
        return encontrado

    return self._buscar_recursivo(dado, no.dir, nivel+1)
  
   

  def remover(self, dado):
   pass
    

  def percorreTudo(self, no = None, nivel = 0):
    #lista todos os dados dos nós, apresentando o nivel em que ele está
    if no is None:
      if self.raiz is None:
        print("Arvore vazia")
      else:
        self.percorreTudo(self.raiz,1)
    else:
      if no.esq is not None:
        self.percorreTudo(no.esq, nivel+1)
      print("-"*nivel, no.dado)
      if no.dir is not None:
        self.percorreTudo(no.dir, nivel+1)


ab01 = Arvore_binaria()
dados = [10,200,7,150,30, 302, 400, 2, 314]
for dado in dados:
  ab01.inserir(dado)

ab01.percorreTudo()
#lista todos os dados dos nós, apresentando o nivel em que ele está
ab01.buscar(302)


'''
ab01.percorreEmOrdem()
ab01.percorrePreOrdem()
ab01.percorrePosOrdem()

'''
```
