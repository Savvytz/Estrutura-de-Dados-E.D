```python
"""
UFRJ - Universidade Federal do Rio de Janeiro
CCMN - Centro de Ciência Matemáticas e da Natureza
IM   - Instituto de Matemática

Disciplina: Estrutura de Dados (MAE014 - ED.2026.2)
Prof.     : Milton R Ramirez (milton@matematica.ufrj.br)

Data Criação  : 01 de outubro de 2026
Data Alteração: 01 de outubro  de 2026

Exemplos de implementação de arvore de grau de saida variavel
sem rotina de inserçao, apenas para trabalhar a recursao na busca
"""

class NoArvore:
  def __init__(self, dado):
    self.dado = dado
    self.filhos = []

  def adicionar_filho(self, filho):
    self.filhos.append(filho)

#arvore FIXA
class Arvore:
  def __init__(self):
    self.raiz = NoArvore(10)
    self.raiz.filhos.append(NoArvore(200))
    self.raiz.filhos.append(NoArvore(7))
    self.raiz.filhos[0].filhos.append(NoArvore(5))
    self.raiz.filhos[0].filhos.append(NoArvore(0))

    self.raiz.filhos[1].filhos.append(NoArvore(13))
    self.raiz.filhos[1].filhos.append(NoArvore(12))
    self.raiz.filhos[1].filhos.append(NoArvore(11))

  def buscaPorValorRecursivo(self, no, valor):
    if no is None:
      return False
    if no.dado == valor:
      return True
    for filho in no.filhos:
      if self.buscaPorValorRecursivo(filho, valor):
        return True
    return False

  def buscaPorValor( self, valor ):
    no = self.raiz
    resultado = self.buscaPorValorRecursivo(no,valor)
    return resultado

#Testes
T = Arvore()
print(T.buscaPorValor(12))
print(T.buscaPorValor(100))

```
